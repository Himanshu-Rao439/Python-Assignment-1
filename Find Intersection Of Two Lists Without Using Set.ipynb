```python
def find_intersection(list1: list,list2: list) -> list:
    """
    Find Common elements between two lists (real-world approach).
    args:
        list1: should be list
        list2: should be list
    returns:
        list
    Raises:
        TypeError: If inputs are not in lists
        ValueError: If lists are empty
    """
    if not isinstance(list1,list) or not isinstance(list2,list):
        raise TypeError("Both arg must be list")
    if not list1 or not list2:
        raise ValueError("Input must not be empty")
    result = []
    for item in list1:
        if item in list2 and item not in result:
            result.append(item)
    return result
#----- MAIN EXECUTION -----
if __name__ == "__main__":
    list1 = [100,2,3,4,5]
    list2 = [4,5,100,200,4]
    try:
        print (find_intersection(list1,list2))
    except (TypeError,ValueError) as e:
        print("Error:",e)
    except Exception as e:
        print("Error:",e)
    
```
